```python
import numpy as np

arr = np.array([[6, -8, 73, -110],
                [np.nan, -8, 0, 94]])

print("Original Array:")
print(arr)

# Replace NaN with 0
arr = np.nan_to_num(arr, nan=0)

print("\nAfter Replacing NaN with 0:")
print(arr)

# Interchange Rows
arr_rows = arr[[1, 0]]

print("\nAfter Interchanging Rows:")
print(arr_rows)

# Interchange Columns
arr_cols = arr[:, [3, 2, 1, 0]]

print("\nAfter Interchanging Columns:")
print(arr_cols)
```

```python
import numpy as np

arr = np.array([[10, np.nan, 30],
                [20, 40, np.nan],
                [30, 50, 60]])

print("Original Array:")
print(arr)

col_mean = np.nanmean(arr, axis=0)

inds = np.where(np.isnan(arr))

arr[inds] = np.take(col_mean, inds[1])

print("\nArray after replacing NaN with column average:")
print(arr)
```

```python
import numpy as np

arr = np.array([10, -5, 20, -15, 30, -2])

print("Original Array:")
print(arr)

arr[arr < 0] = 0

print("\nAfter Replacing Negative Values:")
print(arr)
```

```python
import numpy as np
from scipy import stats

arr1 = np.array([[1, 2, 3],
                 [4, 5, 6]])

arr2 = np.array([[7, 8, 9],
                 [10, 11, 12]])

# Average Array
avg = (arr1 + arr2) / 2

print("Average Array:")
print(avg)

combined = np.concatenate((arr1.flatten(),
                           arr2.flatten()))

print("\nMean =", np.mean(combined))
print("Median =", np.median(combined))
print("Mode =", stats.mode(combined, keepdims=True))
```

```python
#Solve Equation Using linalg() and Inverse Matrix Method
#Equations
# x - 2y + 3z = 9
# -x + 3y - z = -6
# 2x - 5y + 5z = 17


import numpy as np

A = np.array([[1, -2, 3],
              [-1, 3, -1],
              [2, -5, 5]])

B = np.array([9, -6, 17])

# Method 1 : linalg.solve()

solution = np.linalg.solve(A, B)

print("Solution using linalg.solve():")
print(solution)

# Method 2 : Inverse Matrix

A_inv = np.linalg.inv(A)

X = np.dot(A_inv, B)

print("\nSolution using Inverse Matrix:")
print(X)
```
