```python
import numpy as np

a = np.array([[6,-8,73,-110],
              [np.nan,-8,0,94]])

print(a)

a = np.nan_to_num(a,nan=0)

print("After Replacing NaN")
print(a)

a[[0,1],:] = a[[1,0],:]

print("After Row Swap")
print(a)

a[:,[0,1,2,3]] = a[:,[3,2,1,0]]

print("After Column Swap")
print(a)
```

```python
import numpy as np

a = np.array([
    [10,np.nan,30],
    [20,40,np.nan],
    [30,50,60]
])

print(a)

mean = np.nanmean(a,axis=0)

inds = np.where(np.isnan(a))

a[inds] = np.take(mean,inds[1])

print(a)
```

```python
import numpy as np

a = np.array([
    [1,2,3],
    [4,5,6]
])

b = np.array([
    [7,8,9],
    [10,11,12]
])

avg = (a+b)/2

print("Average Array")
print(avg)

c = np.concatenate((a.flatten(),b.flatten()))

print("Mean =",np.mean(c))
print("Median =",np.median(c))
```

```python
from scipy import stats

print(stats.mode(c))
```

```python
import numpy as np

arr1 = np.array([
    [1,-2,3],
    [-1,3,-1],
    [2,-5,5]
])

arr2 = np.array([9,-6,17])

result = np.linalg.solve(arr1,arr2)

print(result)
```

```python
import numpy as np

arr1 = np.array([
    [1,-2,3],
    [-1,3,-1],
    [2,-5,5]
])

arr2 = np.array([9,-6,17])

inv_arr1 = np.linalg.inv(arr1)

result = np.dot(inv_arr1,arr2)

print(result)
```

```python
import numpy as np
import matplotlib.pyplot as plt

subjects = ['Maths','Physics','BEE','OOPs','CS']

sem1 = [75,70,68,72,80]
sem2 = [82,78,75,85,88]

plt.plot(subjects,sem1,
         label='Semester 1',
         color='blue',
         linestyle='--',
         linewidth=2)

plt.plot(subjects,sem2,
         label='Semester 2',
         color='red',
         linestyle='-',
         linewidth=2)

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Semester Result Comparison")

plt.legend()
plt.grid()

plt.show()
```
