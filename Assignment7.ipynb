```python
# 1) Convert a series of date-strings to a timeseries?
import pandas as pd

# Series of date strings
dates = pd.Series(['2026-01-10', '2026-02-15', '2026-03-20'])

print("Original Series:")
print(dates)

# Convert to datetime
timeseries = pd.to_datetime(dates)

print("\nConverted Time Series:")
print(timeseries)
```

```python
# 2)Create two DataFrames, df1 and df2, with a common column (e.g., 'ID'). 
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Mohit', 'Nikhil', 'Preeti', 'Ravi']
})

df2 = pd.DataFrame({
    'ID': [2, 3, 4, 5],
    'Marks': [85, 90, 88, 95]
})

print("DF1")
print(df1)

print("\nDF2")
print(df2)

# >Perform an inner merge on this common column and display the resulting DataFrame.
inner_merge = pd.merge(df1, df2, on='ID', how='inner')

print(inner_merge)

# >Perform a left join of df1 and df2 on the 'ID' column. Explain how missing values are handled in the resulting DataFrame. Right Join and Index-Based Join.
left_join = pd.merge(df1, df2, on='ID', how='left')
print(left_join)

right_join = pd.merge(df1, df2, on='ID', how='right')
print(right_join)

df1_index = df1.set_index('ID')
df2_index = df2.set_index('ID')
joined_df = df1_index.join(df2_index)
print(joined_df)

# >Perform a right join using pd.merge() on a common column, then perform a join using df.join() based on the index. Compare the results. Merging with Multiple Keys.
dfA = pd.DataFrame({
    'ID': [1, 1, 2],
    'City': ['Jaipur', 'Delhi', 'Mumbai'],
    'Name': ['Mohit', 'Nikhil', 'Preeti']
})

dfB = pd.DataFrame({
    'ID': [1, 2, 2],
    'City': ['Jaipur', 'Mumbai', 'Delhi'],
    'Marks': [90, 85, 75]
})

multi_merge = pd.merge(dfA, dfB, on=['ID', 'City'])

print(multi_merge)
```

```python
# 3) Create three DataFrames. Vertically concatenate two of them using pd.concat(), then merge the resulting DataFrame with the third DataFrame on a common key. T Understand join() vs. merge().
df1 = pd.DataFrame({
    'ID': [1, 2],
    'Name': ['Mohit', 'Nikhil']
})

df2 = pd.DataFrame({
    'ID': [3, 4],
    'Name': ['Preeti', 'Ravi']
})

df3 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Marks': [80, 85, 90, 88]
})

df = pd.concat([df1, df2])
print(df)

final_df = pd.merge(df, df3, on='ID')
print(final_df)

```
