```python
# Student Name, Class, Total Marks aur Percentage
# Student Result Program

name = input("Enter Student Name: ")
student_class = input("Enter Class: ")

m1 = float(input("Enter marks of Subject 1: "))
m2 = float(input("Enter marks of Subject 2: "))
m3 = float(input("Enter marks of Subject 3: "))
m4 = float(input("Enter marks of Subject 4: "))
m5 = float(input("Enter marks of Subject 5: "))

total = m1 + m2 + m3 + m4 + m5
percentage = total / 5

print("\n----- RESULT -----")
print("Name:", name)
print("Class:", student_class)
print("Total Marks:", total)
print("Percentage:", percentage, "%")
 

```

```python
# Concatenate Two Strings and Apply String Methods

str1 = input("Enter First String: ")
str2 = input("Enter Second String: ")

new_string = str1 + " " + str2

print("\nConcatenated String:", new_string)

print("lower():", new_string.lower())
print("upper():", new_string.upper())
print("title():", new_string.title())
print("swapcase():", new_string.swapcase())
print("capitalize():", new_string.capitalize())
print("casefold():", new_string.casefold())

print("center(50):", new_string.center(50))

print("count('a'):", new_string.count('a'))
print("endswith('a'):", new_string.endswith('a'))
print("find('a'):", new_string.find('a'))

print("isalnum():", new_string.isalnum())
print("isdigit():", new_string.isdigit())
print("isnumeric():", new_string.isnumeric())
print("isspace():", new_string.isspace())

print("replace('a','@'):", new_string.replace('a', '@'))
```

```python
 # Practice of Assignment Operators

a = 20

print("Initial Value:", a)

a += 5
print("After += :", a)

a -= 3
print("After -= :", a)

a *= 2
print("After *= :", a)

a /= 2
print("After /= :", a)

a //= 2
print("After //= :", a)

a %= 3
print("After %= :", a)

a **= 2
print("After **= :", a)
```

```python
# Student Result with Grade using Conditional Statements

name = input("Enter Student Name: ")
student_class = input("Enter Class: ")

m1 = float(input("Enter marks of Subject 1: "))
m2 = float(input("Enter marks of Subject 2: "))
m3 = float(input("Enter marks of Subject 3: "))
m4 = float(input("Enter marks of Subject 4: "))
m5 = float(input("Enter marks of Subject 5: "))

total = m1 + m2 + m3 + m4 + m5
percentage = total / 5

if percentage >= 60:
    grade = 'A'
elif percentage >= 50:
    grade = 'B'
elif percentage >= 40:
    grade = 'C'
elif percentage >= 33:
    grade = 'D'
else:
    grade = 'F'

print("\n----- RESULT -----")
print("Name:", name)
print("Class:", student_class)
print("Total Marks:", total)
print("Percentage:", percentage, "%")
print("Grade:", grade)
```
